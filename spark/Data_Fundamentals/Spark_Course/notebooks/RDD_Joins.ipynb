```python
from pyspark.sql import SparkSession
spark=(
    SparkSession.builder
    .appName("RDD Transformation")
    .master("local[*]")
    .getOrCreate()
)
sc=spark.sparkContext
print(sc.uiWebUrl)
```

### join()
 - Join is ise when you have two pair RDDs and want to combine records having same key
 - INNER Join - SQL

 
### leftOuterJoin()
### rightOuterJoin()

```python
customers=sc.parallelize([
    ("C101","Vamshi"),
    ("C101","Vijay"),
    ("C101","Ashad"),
    ("C101","Darshan")
])

transections=sc.parallelize([
    ("C101",5000),
    ("C102",3000),
    ("C101",2000),
    ("C105",7000)
])

result=customers.join(transections)
print(result.collect())


# ('C101', ('Vamshi', 5000))
# ('C101', ('Vamshi', 2000))
# ('C101', ('Vijay', 5000))
```
